```python
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
```

```python
dir_path = 'C:/Users/khanh/Downloads/src_cv/img/01/1.jpg'

im = Image.open(dir_path)
plt.imshow(im)
```

```python
im_gray = im.convert('L')
plt.imshow(im_gray, cmap='gray')
```

```python
im_array = np.array(im_gray)
print(im_array)
```

```python
plt.figure()
plt.hist(im_array.flatten(), bins=128)
plt.title('Histogram')
plt.xlabel('Value of pixel')
plt.ylabel('Number')
plt.show()
```
