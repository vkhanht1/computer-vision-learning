```python
from PIL import Image
import matplotlib.pyplot as plt

dir_path = 'C:/Users/khanh/Downloads/src_cv/img/01/1.jpg'

im = Image.open(dir_path)

plt.switch_backend('tkagg')

plt.imshow(im)
plt.title('Click on the image to select image')

points = plt.ginput(5)

print(points)

plt.show()
```

```python
plt.close()

plt.imshow(im)
for point in points:
    x, y = point
    plt.plot(x,y, 'r*')

plt.show()
```
