```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [10,16,20,30,25]

x1 = [1,2,3,4,5]
y1 = [20,15,8,9,20]

plt.plot(x,y, 'bo-', label='Data1')
plt.plot(x1,y1, marker='o',color='r',linestyle='-', label='Data2')
plt.legend()
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
categories=['A','B','C','D','E']
values=[15,10,25,12,18]

plt.bar(categories, values, color='g', alpha=0.6)
plt.xlabel('categories')
plt.ylabel('values')
plt.show()
```

```python
my_labels = ['A','B','C','D']
size = [15,30,45,10]
my_colors = ['red','green','blue','yellow']

plt.pie(size,labels=my_labels,colors=my_colors, autopct='%1.1f%%')
plt.title('chart')
plt.show()
```
